```python
import pandas as pd

df = pd.read_csv('free_tweet_export.csv')  # Modifier le chemin si nécessaire
df.head()
```

```python
# Créer le dataset nettoyé
df['id'] = df['id'].astype('string')

df['in_reply_to'] = df['in_reply_to'].astype('string')
df.info()
```

```python
# Compute count
id_set = df['id']
matches = df['in_reply_to'].isin(id_set)
count_matches = matches.sum()
count_matches
```
