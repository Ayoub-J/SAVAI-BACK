## 2) Filtrage

#### -Ne garder que les tweets pertinents SAV

```python
# Supprimer les lignes où la colonne 'name' est 'Free' ou 'Free 1337'
df = df[~df["screen_name"].isin(["free", "Free_1337", "Freebox","FreeFoot","FreenewsActu", "UniversFreebox"])]

print("Taille après filtres:", df.shape)
```

```python
df = df.drop_duplicates(subset='id')
```

```python
#On garde donc toutes les lignes qui ne sont pas des retweets
is_rt = df['full_text'].astype(str).str.startswith('RT @')
if 'retweeted_status' in df.columns:
    is_rt = is_rt | df['retweeted_status'].notna()
```

```python
df = df[~is_rt]
print(df.shape)
```

```python
#Filtre “candidat SAV” par mots-clés
sav_keywords = [
    'panne','réseau','connexion','internet','débit','fibre','adsl',
    'facture','prélèvement','abonnement','forfait','résiliation',
    'sav','assistance','support','technicien','incident','ticket'
]
pat = '|'.join(sav_keywords)
df['is_sav_candidate'] = df['full_text'].astype(str).str.lower().str.contains(pat, regex=True)
df = df[df['is_sav_candidate'] == True].copy()

print("Taille après filtres:", df.shape)
```

## 3) Nettoyage du texte

```python
import re, emoji, html

def clean_tweet(t):
    t = str(t)
    t = html.unescape(t)  # corrige les caractères encodés (ex : &amp;)
    t = t.lower()

    # 1️⃣ Retirer les retours à la ligne et caractères invisibles
    t = re.sub(r'[\r\n\t]', ' ', t)

    # 2️⃣ Supprimer les URLs
    t = re.sub(r'http\S+|www\S+|pic\.twitter\.com\S+', ' ', t)

    # 3️⃣ Supprimer les mentions @
    t = re.sub(r'@\w+', ' ', t)

    # 4️⃣ Supprimer les hashtags
    t = re.sub(r'#', ' ', t)

    # 5️⃣ Supprimer les emojis
    t = emoji.replace_emoji(t, replace=' ')

    # 6️⃣ Supprimer les caractères non alphabétiques
    t = re.sub(r'[^a-zàâçéèêëîïôûùüÿñæœ0-9!?.,;: ]', ' ', t)

    # 7️⃣ Réduire les espaces multiples
    t = re.sub(r'\s+', ' ', t).strip()

    return t

# Réappliquer le nettoyage
df['clean_text'] = df['full_text'].apply(clean_tweet)


# petites features utiles
df['nb_mots'] = df['clean_text'].str.split().str.len()
df['nb_exclam'] = df['full_text'].str.count('!')
df['nb_interro'] = df['full_text'].str.count(r'\?')
df['nb_emoji'] = df['full_text'].apply(lambda x: sum(c in emoji.EMOJI_DATA for c in str(x)))
df['has_media'] = df.get('media', []).astype(str).str.len().gt(2) if 'media' in df.columns else False
```

## 4) Enrichissements

### 4.1 Sentiment (baseline simple, règles)

```python
neg_words = ['panne','coupure','marche pas','bug','injoignable','problème','problème','nul','scandale','attente','remboursement','réclame','facture','bloqué','impossible']
pos_words = ['merci','top','parfait','super','bravo','nickel','génial']

def rule_sentiment(txt):
    t = str(txt)
    if any(w in t for w in neg_words): return 'negatif'
    if any(w in t for w in pos_words): return 'positif'
    return 'neutre'

df['sentiment_lex'] = df['clean_text'].apply(rule_sentiment)
```

### 4.2 Thèmes (regex/mots-clés)

```python
import numpy as np

theme_map = {
    'reseau': ['panne','réseau','connexion','débit','fibre','adsl','coupure','box','freebox','ping','latence'],
    'facture': ['facture','prélèvement','montant','forfait','paiement','surfacturation'],
    'abonnement': ['abonnement','résiliation','portabilité','option','engagement'],
    'support': ['sav','assistance','support','technicien','ticket','incident']
}

def detect_theme(t):
    t = str(t)
    for theme, kws in theme_map.items():
        if any(k in t for k in kws):
            return theme
    return 'autre'

df['theme_principal'] = df['clean_text'].apply(detect_theme)
```

### 4.3 Urgence

```python
urg_markers = [
    'aucun accès','aucun acces','plus d\'internet','plus internet',
    'impossible de','depuis hier','depuis 2 jours','depuis 3 jours','ça fait','ca fait','urgent','bloqué','bloque'
]

def urgency_flag(t):
    t = str(t)
    return int(any(m in t for m in urg_markers))

df['urgence_flag'] = df['clean_text'].apply(urgency_flag)
```
