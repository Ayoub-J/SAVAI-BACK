```python
df_path = "C:/Users/AyoubJELLOULI/Desktop/HETIC/Projet_RNCP/git_repos/SAVAI-BACK/free_tweet_export.csv"
df = pd.read_csv(df_path)
df.head()
```

```python
# Sélectionner uniquement les colonnes souhaitées
df = df[["id", "created_at", "full_text", "screen_name", "name", "user_id", "in_reply_to", "retweeted"]].copy()
df
```

```python
df.info()
```

```python
df['created_at'] = pd.to_datetime(df['created_at'], utc=True)
df.info()
```

```python
df.head()
```

```python
df['created_at_local'] = df['created_at'].dt.tz_convert('Europe/Paris')
df.head()
```

```python
df.info()
```

```python
df['full_text'] = df['full_text'].astype('string')
df['screen_name'] = df['screen_name'].astype('string')
df['name'] = df['name'].astype('string')
df['user_id'] = df['user_id'].astype('Int64')
df['in_reply_to'] = df['in_reply_to'].astype('Int64')
df.info()

```
